# Week 12 Debugging Demo
### Reading errors, print(), %debug, and assert

Debugging means finding and fixing the cause of a problem in your code. Bugs come in three kinds, and each kind has a tool that fits.

| Kind of bug | What happens | Tool |
|-------------|--------------|------|
| Syntax error | Python refuses to run the cell | Read the error message |
| Runtime error | The code starts, then crashes with a message | Read the message, then `%debug` |
| Logic error | The code runs but gives the wrong answer | `print()` to trace it, `assert` to lock in the fix |

**How to use this notebook:** Run the cells one at a time, from the top. Most demo cells are meant to fail or give a wrong answer until you fix them. Run the setup cell below before Demo B and Demo D.

## Setup
Run this cell first. It defines a function that Demo B and Demo D both use.

````python
def average_expense(expenses):
    total = 0
    for amount in expenses:
        total += amount
    return total / len(expenses)
````

## 1. Reading an error message

### Demo A: Syntax error
Python checks the whole cell before it runs anything. Run the cell, then read the message. The caret (^) points to where Python got confused.

````python
# Demo A: Syntax error
expense = 45.50
if expense > 40
    print("Large expense")
````

### Demo B: Name error
Run the cell and read the last line of the message first. Then compare the name in the message with the name you created.

````python
# Demo B: Name error
monthly_expenses = [120.0, 45.5, 80.0, 60.5]
print("Average expense:", average_expense(monthly_expense))
````

**How to read a traceback, from the bottom up**
- The **last line** says what went wrong.
- The **line number and arrow** say where.
- The lines above show the path Python took to get there.

## 2. Logic error and `print()`

### Demo C: The code runs, but the answer is wrong
This function should add income to the balance and subtract expenses. Before you run it, work out the right answer: starting with 100, then 500 of income, then expenses of 120 and 80.

Run the cell. Does the output match your answer? If not, add `print()` lines inside the loop to show each transaction and the running balance.

````python
# Demo C: Logic error
def calculate_balance(starting_balance, transactions):
    balance = starting_balance
    for transaction in transactions:
        if transaction["transaction_type"] == "income":
            balance += transaction["amount"]
        else:
            balance += transaction["amount"]
    return balance

transactions = [
    {"transaction_type": "income", "amount": 500},
    {"transaction_type": "expense", "amount": 120},
    {"transaction_type": "expense", "amount": 80},
]

print("Ending balance:", calculate_balance(100, transactions))
````

### Using `%debug`
After a cell raises an error, run `%debug` in the next cell. It opens the debugger on the line that failed. In VS Code, the prompt appears as an input bar at the **top of the window**.

`%debug` is **post-mortem**: the code has already stopped, so you can look around but you cannot run it forward.

| Command | What it does |
|---------|--------------|
| `p variable` | Prints a value (example: `p total`) |
| `pp variable` | Pretty-prints a value, handy for lists and dictionaries |
| `l` | Lists the code around the current line |
| `w` | Shows where you are in the chain of function calls |
| `u` / `d` | Moves up or down one level in that chain |
| `q` | Quits the debugger |

Press **Enter** after each command. When you are finished, type `q` and press Enter.

The commands `n`, `s`, and `c` do not step or continue in `%debug`. They end the session.

If the input bar disappears, click the **Interrupt** button in the notebook toolbar.

The debugger session is **not saved** in your notebook output. When you need to show what you found, write it in a Markdown cell.

## 4. Unit tests and regression tests with `assert`
An `assert` statement checks that something is true. If it is true, nothing happens. If it is false, Python raises an `AssertionError`.

A check of one small piece of code, such as one function, using values where you already know the right answer, is called a **unit test**. Each `assert` below is a unit test. Try a normal case and some edge cases, such as an empty list, because edge cases are where bugs hide.

Run the cell below **before** you fix Demo C. It should fail. Then fix Demo C, rerun the Demo C cell, and run this cell again. It should run with no output.

A unit test you keep so a bug cannot sneak back in is called a **regression test**. Keep your asserts in a cell like this one, and rerun them after every change to the function.

````python
# Unit tests for calculate_balance
# Normal case
assert calculate_balance(100, transactions) == 400, "Expected 400"
# Edge case: no transactions
assert calculate_balance(0, []) == 0, "An empty list should leave the balance unchanged"
# Edge case: an expense equal to the whole balance
assert calculate_balance(50, [{"transaction_type": "expense", "amount": 50}]) == 0, "Expected 0"
````
