## Exercise 3: Inspect a Crash with `%debug`
This function adds up the transactions of one type. One transaction in the list is causing a crash.

1. Run the cell below and read the message.
2. Run the `%debug` cell right after it. In VS Code, the prompt appears as an input bar at the top of the window.
3. Use `p transaction` to see the record the loop was working on when it crashed. Type `q` and press Enter to leave the debugger.
4. Fix the problem, then rerun the cell and the check below it.

The debugger session is **not saved** in your notebook, so write down what you found in your answer.

```python
# Exercise 3
def total_by_type(transactions, wanted_type):
    total = 0
    for transaction in transactions:
        if transaction["transaction_type"] == wanted_type:
            total += transaction["amount"]
    return total

transactions = [
    {"transaction_type": "income", "amount": 2500},
    {"transaction_type": "expense", "amount": 120},
    {"transaction_type": "expense", "Amount": 45},
    {"transaction_type": "expense", "amount": 80},
]

print("Total expenses:", total_by_type(transactions, "expense"))
```
